```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense

print("Libraries imported successfully!")


# STEP 2: Load Fashion MNIST dataset
print("\nSTEP 2: Loading Fashion MNIST dataset...")

(x_train, _), (x_test, _) = fashion_mnist.load_data()

print("Original training shape:", x_train.shape)
print("Original testing shape:", x_test.shape)


# STEP 3: Normalize images
print("\nSTEP 3: Normalizing image data...")

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

print("Normalization completed!")
print("Minimum pixel value:", x_train.min())
print("Maximum pixel value:", x_train.max())


# STEP 4: Flatten images
print("\nSTEP 4: Flattening images...")

x_train = x_train.reshape(-1, 784)
x_test = x_test.reshape(-1, 784)

print("Training data:", x_train.shape)
print("Testing data :", x_test.shape)


# STEP 5: Add noise to images
print("\nSTEP 5: Adding noise to images...")

noise_factor = 0.3

np.random.seed(42)

x_train_noisy = x_train + noise_factor * np.random.normal(
    0, 1, x_train.shape
)

x_test_noisy = x_test + noise_factor * np.random.normal(
    0, 1, x_test.shape
)

x_train_noisy = np.clip(x_train_noisy, 0., 1.)
x_test_noisy = np.clip(x_test_noisy, 0., 1.)

print("Noise added successfully!")
print("Noisy training data:", x_train_noisy.shape)
print("Noisy testing data :", x_test_noisy.shape)


# STEP 6: Build Denoising Autoencoder
print("\nSTEP 6: Building Denoising Autoencoder...")

input_img = Input(shape=(784,))

# Encoder
encoded = Dense(128, activation="relu")(input_img)
encoded = Dense(64, activation="relu")(encoded)
encoded = Dense(32, activation="relu")(encoded)

# Decoder
decoded = Dense(64, activation="relu")(encoded)
decoded = Dense(128, activation="relu")(decoded)
decoded = Dense(784, activation="sigmoid")(decoded)

# Complete Autoencoder
autoencoder = Model(input_img, decoded)

print("Denoising Autoencoder created successfully!")
autoencoder.summary()


# STEP 7: Compile model
print("\nSTEP 7: Compiling model...")

autoencoder.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mse"]
)

print("Model compiled successfully!")


# STEP 8: Configure Early Stopping
print("\nSTEP 8: Configuring Early Stopping...")

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

print("Early Stopping configured!")


# STEP 9: Train Denoising Autoencoder
print("\nSTEP 9: Training Denoising Autoencoder...")

history = autoencoder.fit(
    x_train_noisy,
    x_train,
    epochs=20,
    batch_size=256,
    shuffle=True,
    validation_split=0.1,
    callbacks=[early_stopping],
    verbose=1
)

print("Training completed!")
print("Epochs completed:", len(history.history["loss"]))


# STEP 10: Evaluate model
print("\nSTEP 10: Evaluating model...")

test_loss, test_mse = autoencoder.evaluate(
    x_test_noisy,
    x_test,
    verbose=0
)

print("Test Loss:", test_loss)
print("Test MSE :", test_mse)


# STEP 11: Plot Loss and MSE
print("\nSTEP 11: Plotting Loss and MSE graphs...")

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.title("Fashion MNIST Denoising Autoencoder Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)

plt.plot(history.history["mse"], label="Training MSE")
plt.plot(history.history["val_mse"], label="Validation MSE")

plt.title("Fashion MNIST Denoising Autoencoder MSE")
plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()


# STEP 12: Reconstruct noisy test images
print("\nSTEP 12: Reconstructing noisy images...")

decoded_imgs = autoencoder.predict(
    x_test_noisy,
    verbose=0
)

print("Reconstructed images shape:", decoded_imgs.shape)


# STEP 13: Display noisy, denoised, and original images
print("\nSTEP 13: Displaying reconstruction results...")

plt.figure(figsize=(15, 5))

for i in range(10):

    # Noisy image
    plt.subplot(3, 10, i + 1)
    plt.imshow(x_test_noisy[i].reshape(28, 28), cmap="gray")
    plt.axis("off")

    # Denoised image
    plt.subplot(3, 10, i + 11)
    plt.imshow(decoded_imgs[i].reshape(28, 28), cmap="gray")
    plt.axis("off")

    # Original image
    plt.subplot(3, 10, i + 21)
    plt.imshow(x_test[i].reshape(28, 28), cmap="gray")
    plt.axis("off")

plt.suptitle(
    "Fashion MNIST: Noisy → Denoised → Original"
)
plt.tight_layout()
plt.show()


# STEP 14: Calculate reconstruction MSE
print("\nSTEP 14: Calculating reconstruction error...")

reconstruction_mse = np.mean(
    np.square(x_test - decoded_imgs)
)

print("Reconstruction MSE:", reconstruction_mse)
```
