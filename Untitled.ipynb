```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.regularizers import l1

print("Libraries imported successfully!")


# STEP 2: Load MNIST dataset
print("\nSTEP 2: Loading MNIST dataset...")

(x_train, _), (x_test, _) = mnist.load_data()

print("Original training shape:", x_train.shape)
print("Original testing shape:", x_test.shape)


# STEP 3: Normalize images
print("\nSTEP 3: Normalizing image data...")

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

print("Normalization completed!")
print("Minimum pixel value:", x_train.min())
print("Maximum pixel value:", x_train.max())


# STEP 4: Flatten images
print("\nSTEP 4: Flattening images...")

x_train = x_train.reshape(-1, 784)
x_test = x_test.reshape(-1, 784)

print("Training data:", x_train.shape)
print("Testing data :", x_test.shape)


# STEP 5: Build Sparse Autoencoder
print("\nSTEP 5: Building Sparse Autoencoder...")

input_img = Input(shape=(784,))

# Encoder
encoded = Dense(128, activation="relu")(input_img)
encoded = Dense(64, activation="relu")(encoded)

# Sparse bottleneck layer using L1 regularization
encoded = Dense(
    32,
    activation="relu",
    activity_regularizer=l1(1e-5)
)(encoded)

# Decoder
decoded = Dense(64, activation="relu")(encoded)
decoded = Dense(128, activation="relu")(decoded)
decoded = Dense(784, activation="sigmoid")(decoded)

# Complete Autoencoder
autoencoder = Model(inputs=input_img, outputs=decoded)

# Separate Encoder
encoder = Model(inputs=input_img, outputs=encoded)

print("Sparse Autoencoder created successfully!")
autoencoder.summary()


# STEP 6: Compile model
print("\nSTEP 6: Compiling model...")

autoencoder.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mse"]
)

print("Model compiled successfully!")


# STEP 7: Configure Early Stopping
print("\nSTEP 7: Configuring Early Stopping...")

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

print("Early Stopping configured!")


# STEP 8: Train Sparse Autoencoder
print("\nSTEP 8: Training Sparse Autoencoder...")

history = autoencoder.fit(
    x_train,
    x_train,
    epochs=20,
    batch_size=256,
    shuffle=True,
    validation_split=0.1,
    callbacks=[early_stopping],
    verbose=1
)

print("Training completed!")
print("Epochs completed:", len(history.history["loss"]))


# STEP 9: Evaluate model
print("\nSTEP 9: Evaluating model...")

test_loss, test_mse = autoencoder.evaluate(
    x_test,
    x_test,
    verbose=0
)

print("Test Loss:", test_loss)
print("Test MSE :", test_mse)


# STEP 10: Plot Loss and MSE
print("\nSTEP 10: Plotting Loss and MSE graphs...")

plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.title("Sparse Autoencoder Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)

plt.plot(history.history["mse"], label="Training MSE")
plt.plot(history.history["val_mse"], label="Validation MSE")

plt.title("Sparse Autoencoder MSE")
plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()


# STEP 11: Reconstruct test images
print("\nSTEP 11: Reconstructing test images...")

decoded_imgs = autoencoder.predict(
    x_test,
    verbose=0
)

print("Reconstructed images shape:", decoded_imgs.shape)


# STEP 12: Generate encoded representations
print("\nSTEP 12: Generating compressed representations...")

encoded_imgs = encoder.predict(
    x_test[:10],
    verbose=0
)

print("Original image dimensions:", x_test.shape[1])
print("Encoded image dimensions:", encoded_imgs.shape[1])

print("Compression completed!")


# STEP 13: Display original and reconstructed images
print("\nSTEP 13: Displaying original and reconstructed images...")

plt.figure(figsize=(15, 4))

for i in range(10):

    # Original image
    plt.subplot(2, 10, i + 1)
    plt.imshow(x_test[i].reshape(28, 28), cmap="gray")
    plt.axis("off")

    # Reconstructed image
    plt.subplot(2, 10, i + 11)
    plt.imshow(decoded_imgs[i].reshape(28, 28), cmap="gray")
    plt.axis("off")

plt.suptitle("Original Images (Top) vs Reconstructed Images (Bottom)")
plt.tight_layout()
plt.show()


# STEP 14: Calculate reconstruction MSE
print("\nSTEP 14: Calculating reconstruction error...")

reconstruction_mse = np.mean(
    np.square(x_test - decoded_imgs)
)

print("Reconstruction MSE:", reconstruction_mse)

```
